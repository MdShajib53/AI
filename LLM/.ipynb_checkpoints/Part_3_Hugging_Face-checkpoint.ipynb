<h1 style='color:green'>Transformer Pipeline</h1>

# Input -> Tokenizer -> Model -> Post Processing -> Output

```python
from transformers import pipeline
```

```python
sentiment_classifier = pipeline("sentiment-analysis")
```

```python
sentiment_classifier("I'm so excited to be learnig about large language models")
```

```python
ner = pipeline("ner", model="dslim/bert-base-NER")
```

```python
ner("Her name is Anna and she works in New York City for Morgan Stanley")
```

```python
zeroshot_classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
```

```python
sequence_to_classify = "one day I will see the world"
candidate_labels = ['travel', 'cooking', 'dancing']
```

```python
zeroshot_classifier(sequence_to_classify, candidate_labels)
```

# Pre-trained Tokenizers

```python
from transformers import AutoTokenizer
```

```python
model = "bert-base-uncased"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
sentence1 = "I'm so excited to be learning about large language models."
sentence2 = "I'm super exited about tomorrow iftar party."
```

```python
input_ids = tokenizer(sentence1)
print(input_ids)
```

```python
input_ids = tokenizer(sentence1, sentence2)
print(input_ids)
```

```python
tokens = tokenizer.tokenize(sentence1)
```

```python
tokens
```

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
```

```python
decoded_ids = tokenizer.decode(token_ids)
print(decoded_ids)
```

```python
tokenizer.decode(101)
```

```python
tokenizer.decode(102)
```

```python
model2 = "xlnet-base-cased"
```

```python
tokenizer2 = AutoTokenizer.from_pretrained(model2)
```

```python
input_ids = tokenizer2(sentence1)
```

```python
print(input_ids)
```

```python
tokens = tokenizer2.tokenize(sentence1)
print(tokens)
```

```python
token_ids = tokenizer2.convert_tokens_to_ids(tokens)
print(token_ids)
```

```python
tokenizer2.decode(4)
```

```python
tokenizer2.decode(3)
```

<h1 style='color:green'>Hugging Face and Pytorch/Tensorflow</h1>

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
print(sentence1)
print(input_ids)
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
input_ids_pt = tokenizer(sentence1, return_tensors="pt")
print(input_ids_pt)
```

```python
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
with torch.no_grad(): 
    logits = model(**input_ids_pt).logits
predicted_class_id = logits.argmax().item()
model.config.id2label[predicted_class_id]
```

<h1 style='color:green'>Saving and Loading Models</h1>

```python
model_directory = "my_saved_models"
```

```python
tokenizer.save_pretrained(model_directory)
```

```python
model.save_pretrained(model_directory)
```

```python
my_tokenizer = AutoTokenizer.from_pretrained(model_directory)
```

```python
my_model = AutoModelForSequenceClassification.from_pretrained(model_directory)
```

```python
with torch.no_grad(): 
    logits = my_model(**input_ids_pt).logits
predicted_class_id = logits.argmax().item()
my_model.config.id2label[predicted_class_id] 
```
